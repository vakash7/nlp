```python
# Define the positive and negative keyword lists
positive_keywords = ["amazing", "heartwarming", "must-watch", "brilliant", "masterpiece"]
negative_keywords = ["boring", "forced", "weak", "regret", "disappointing"]

# Test dataset (unlabeled)
test_reviews = [
    "I found the movie heartwarming but a bit too long.",
    "The story was confusing and the pacing was terrible.",
    "A brilliant and must-watch film with a strong message.",
    "Just okay, nothing special or memorable.",
    "An amazing performance, but the script felt forced."
]

# Function to classify sentiment
def classify_review(review):
    review_lower = review.lower()
    pos_count = sum(1 for word in positive_keywords if word in review_lower)
    neg_count = sum(1 for word in negative_keywords if word in review_lower)

    if pos_count >= 2:
        return "Positive"
    elif neg_count >= 2:
        return "Negative"
    elif pos_count > neg_count:
        return "Slightly Positive"
    elif neg_count > pos_count:
        return "Slightly Negative"
    else:
        return "Neutral"

# Predict sentiments for the test dataset
predicted_outputs = [classify_review(review) for review in test_reviews]

# Print predictions
for i, (review, prediction) in enumerate(zip(test_reviews, predicted_outputs), 1):
    print(f"Review {i}: \"{review}\"\nPredicted Sentiment: {prediction}\n")
```
